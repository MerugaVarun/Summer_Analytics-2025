### Find the car (or cars) with the best ratio of horsepower to weight among all cars that also have above-median mpg.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
Cars = pd.read_csv('cars.csv')
median_mpg = Cars['mpg'].median()

above_median_mpg_cars = Cars[Cars['mpg'] > median_mpg]

above_median_mpg_cars['hp_to_weight'] = above_median_mpg_cars['horsepower'] / above_median_mpg_cars['weight']

best_ratio_car = above_median_mpg_cars.loc[above_median_mpg_cars['hp_to_weight'].idxmax()]

print("Car(s) with the best horsepower-to-weight ratio and above-median mpg:\n", best_ratio_car)

average_mpg_per_year = Cars.groupby('model_year')['mpg'].mean()

plt.figure(figsize=(10, 6))
sns.lineplot(x=average_mpg_per_year.index, y=average_mpg_per_year.values)
plt.title('Evolution of Average MPG Over the Years')
plt.xlabel('Year')
plt.ylabel('Average MPG')
plt.show()
```

### Design a multi-line plot using Matplotlib or Seaborn that shows the evolution of average mpg over the years, separately for each origin

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

mpg_df = sns.load_dataset('mpg')

avg_mpg_df = mpg_df.groupby(['model_year', 'origin'])['mpg'].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.lineplot(x='model_year', y='mpg', hue='origin', data=avg_mpg_df)

plt.xlabel('Year')
plt.ylabel('Average MPG')
plt.title('Evolution of Average MPG by Origin')

plt.show()
```

### Create a Seaborn scatterplot (or PairGrid) where:

X = horsepower

Color by: origin

Hue order = ['japan', 'europe', 'usa']

Add meaningful plot titles and axis titles.

### We define a “consistent” car model as one that was produced over multiple years and had very low variation in mpg across those years (standard deviation < 1.0).

Tasks:

Identify car names that appear in more than one model_year.

For each such name, compute the standard deviation of mpg across years.

Return the car(s) with the lowest variation in mpg, among those with at least 2 appearances and std(mpg) < 1.0.

Report the model name(s), number of appearances, and the average mpg.

Bonus:
Sort the result by number of appearances (descending), then mpg (descending).

```python
import pandas as pd
df = pd.read_csv('Cars.csv')

name_year_counts = df.groupby('name')['model_year'].nunique()
repeated_models = name_year_counts[name_year_counts > 1].index

filtered_df = df[df['name'].isin(repeated_models)]

model_stats = filtered_df.groupby('name').agg(
    mpg_std=('mpg', 'std'),
    count=('mpg', 'count'),
    avg_mpg=('mpg', 'mean')
)

consistent_models = model_stats[model_stats['mpg_std'] < 1.0]

consistent_sorted = consistent_models.sort_values(by=['count', 'avg_mpg'], ascending=[False, False])

consistent_sorted.reset_index(inplace=True)

print(consistent_sorted)
```
